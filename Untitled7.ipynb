```python
# ==========================================
# 1. CLASE BASE: FiguraGeometrica
# ==========================================
class FiguraGeometrica:
    def __init__(self, ancho, alto):
        # Se asignan a través de los setters para validar desde el inicio
        self.ancho = ancho
        self.alto = alto

    @property
    def ancho(self):
        return self._ancho

    @ancho.setter
    def ancho(self, valor):
        # Validación: mayor que 0
        if valor > 0:
            self._ancho = valor
        else:
            raise ValueError(f"Ancho inválido ({valor}): Debe ser mayor a 0")

    @property
    def alto(self):
        return self._alto

    @alto.setter
    def alto(self, valor):
        # Validación: mayor que 0
        if valor > 0:
            self._alto = valor
        else:
            raise ValueError(f"Alto inválido ({valor}): Debe ser mayor a 0")

    def area(self):
        # Implementación base: ancho * alto
        return self.ancho * self.alto

    def perimetro(self):
        # Según instrucción: no implementarlo en la base
        pass

    def __str__(self):
        return f"FiguraGeometrica [Ancho: {self.ancho}, Alto: {self.alto}]"


# ==========================================
# 2. CLASE HIJA: Cuadrado
# ==========================================
class Cuadrado(FiguraGeometrica):
    def __init__(self, lado):
        # Regla: Recibe un solo valor y crea ambos iguales usando setters
        super().__init__(lado, lado)

    def area(self):
        # Sobrescritura del método area
        return self.ancho * self.alto

    def perimetro(self):
        # Sobrescritura del método perimetro
        return self.ancho * 4

    def __str__(self):
        return f"Cuadrado [Lado: {self.ancho}]"


# ==========================================
# 3. CLASE HIJA: Rectangulo
# ==========================================
class Rectangulo(FiguraGeometrica):
    def __init__(self, ancho, alto):
        super().__init__(ancho, alto)

    def area(self):
        # Sobrescritura del método area
        return self.ancho * self.alto

    def perimetro(self):
        # Sobrescritura del método perimetro
        return (self.ancho + self.alto) * 2

    def __str__(self):
        return f"Rectángulo [Base: {self.ancho}, Altura: {self.alto}]"


# ==========================================
# FUNCIONES GLOBALES DE CÁLCULO
# ==========================================
def sumar_areas(figuras: list):
    """Acumula y devuelve el total de las áreas de una lista de figuras."""
    total = 0
    for figura in figuras:
        # Polimorfismo: invoca area() sin saber el tipo específico
        total += figura.area()
    return total

def sumar_perimetros(figuras: list):
    """Acumula y devuelve el total de los perímetros de una lista de figuras."""
    total = 0
    for figura in figuras:
        total += figura.perimetro()
    return total


# ==========================================
# 4. PROGRAMA PRINCIPAL (main.py)
# ==========================================
if __name__ == "__main__":
    print("--- INICIO DEL PROGRAMA FIGURAS GEOMÉTRICAS ---")

    try:
        # 1. Crear dos cuadrados y dos rectángulos
        c1 = Cuadrado(lado=5)
        c2 = Cuadrado(lado=10)
        r1 = Rectangulo(ancho=8, alto=4)
        r2 = Rectangulo(ancho=12, alto=6)

        lista_figuras = [c1, c2, r1, r2]

        # 2. Mostrar datos iniciales, área y perímetro
        print("\n1. Datos iniciales e Impresión de objetos:")
        for fig in lista_figuras:
            print(f"-> {fig} | Área: {fig.area()} | Perímetro: {fig.perimetro()}")

        # 3. Modificación de valores para demostrar encapsulamiento
        print("\n2. Modificación de valores (Encapsulamiento):")
        print(f"Rectángulo 1 original: {r1}")
        r1.ancho = 15  # Se ejecuta el @setter
        r1.alto = 7    # Se ejecuta el @setter
        print(f"Rectángulo 1 modificado: {r1}")

        # 4. Uso de funciones globales para sumar totales
        print("\n3. Resultados Totales de la lista:")
        area_total = sumar_areas(lista_figuras)
        perimetro_total = sumar_perimetros(lista_figuras)

        print(f"Suma total de todas las ÁREAS: {area_total}")
        print(f"Suma total de todos los PERÍMETROS: {perimetro_total}")

        # 5. Demostración de validación (Levantar ValueError)
        print("\n4. Prueba de Validación (Valor inválido):")
        error_test = Cuadrado(lado=-2) # Esto disparará el raise ValueError

    except ValueError as e:
        print(f"ERROR DETECTADO: {e}")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")
```
